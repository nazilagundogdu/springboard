# User Adoption Prediction Report

## Objective

The goal of this analysis was to identify which factors predict whether a user becomes an **adopted user**, defined as a user who has logged into the product on at least three separate days within one seven-day period.

## Data

- **User Table:** Contains user demographics and metadata such as signup source, marketing preferences, and invitation details.
- **User Engagement Table:** Daily login records for each user.

## Approach

1. **Feature Engineering:**  
   Created binary target variable indicating adoption status. Extracted features from user data such as:
   - Invitation status (`invited`)
   - Mailing list opt-in
   - Marketing drip enablement
   - Signup source (personal projects, org invite, Google auth, etc.)

2. **Modeling:**  
   Built and compared multiple classification models to predict adoption:
   - Model 1: Used all features
   - Model 2: Reduced feature set

3. **Evaluation Metrics:**  
   Compared models on precision, recall, F1-score, and accuracy. Due to imbalanced data, focused on minority class (adopted users) recall and F1.

4. **Additional Metrics:**  
   Compared ROC AUC and Precision-Recall curves for a fuller picture of model performance.

## Key Findings

| Feature                          | Impact on Adoption                                            |
|---------------------------------|--------------------------------------------------------------|
| `invited`                       | Strong **negative** predictor — invited users less likely to adopt |
| `opted_in_to_mailing_list`      | Positive predictor — users opting in more likely to adopt    |
| `creation_source_SIGNUP`        | Negative — self-signups less likely to adopt                  |
| `creation_source_ORG_INVITE`    | Negative — organization invites linked to lower adoption     |
| `enabled_for_marketing_drip`    | Positive but modest effect                                    |

### Model Comparison

| Metric              | Model 1 (All Features) | Model 2 (Reduced Features) | Recommendation      |
|---------------------|-----------------------|----------------------------|---------------------|
| Accuracy            | 0.87                  | 0.87                       | Tie                 |
| Precision (Adopted) | 0.95                  | 0.96                       | Slight edge Model 2  |
| Recall (Adopted)    | 0.21                  | 0.20                       | Model 1 better      |
| F1-score (Adopted)  | 0.35                  | 0.34                       | Model 1 better      |
| ROC AUC             | Comparable (~0.75)    | Comparable (~0.75)          | Tie                 |
| Precision-Recall AP  | Comparable            | Comparable                 | Tie                 |

Model 1 demonstrates slightly better recall and F1 for the critical adopted user class, making it preferable for identifying potential adopters, despite very similar overall metrics.

## Recommendations and Next Steps

- **Model selection:** Prefer Model 1 for balanced detection of adopters.
- **Business focus:** Since catching adopters is important, prioritize recall over precision if missing adopters is costly.
- **Further tuning:** Explore threshold adjustment, additional features, or sampling techniques to improve minority class prediction.
- **Additional data:** Including more behavioral or engagement features could enhance predictions.
- **Visualization:** ROC and Precision-Recall curves validate models’ discriminative ability and help choose operating points.

# Code

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix
from sklearn.metrics import ConfusionMatrixDisplay
from sklearn.metrics import classification_report
from sklearn.inspection import permutation_importance
import warnings
warnings.filterwarnings("ignore")
```

```python
df1 = pd.read_csv('takehome_users.csv',encoding='ISO-8859-1')
```

```python
df1
```

```python
df2 = pd.read_csv('takehome_user_engagement.csv')
df2.head()
```

We need to join the two tables. They both share ids for the users, which are distinctive. So, we can use that column as the shared column between the two. First, we need to rename the column in one of the dataframes to match the other one.  

```python
df1.rename(columns = {'object_id' : 'user_id'}, inplace = True)
df1.head()
```

```python
df = pd.merge(df1, df2, on = 'user_id', how = 'inner')
df.head()
```

Now, we need to group by a specific week and sum up the visit counts for that week. First, we need to check the time range and then aggregate the data based on that. Even before that, we need to make sure the data type for this column is a datetime object. 

```python
df.info()
```

The data type is incorrect. Let's fix that. And while we are at it, fix other issues with the data.

**1. dtypes**

**1.1. `datetime` objects**

```python
df['creation_time'] = pd.to_datetime(df['creation_time'])
df['time_stamp'] = pd.to_datetime(df['time_stamp'])
df.info()
```

# last_session_creation_time?

**1.2. Values**

```python
df['opted_in_to_mailing_list'].unique() # correct values
```

```python
df['enabled_for_marketing_drip'].unique() # correct values
```

**1.3. `null` objects**

```python
df.isnull().sum()
```

Since the `invited_by_user_id` is not a given, and the users might have joined without an invitation, the `null` values in this columns are expected. We'll leave the column as is for the time being. 

**2. Aggregating**

```python
# aggregating on sum of `visited` based on 1 week intervals and keeping the categorical columns 
df_agg = df.groupby([pd.Grouper(key='time_stamp', freq='1W'), 'user_id', 'creation_source']).sum('visited').reset_index()
```

```python
df_agg.head()
```

```python
df_agg.info()
```

```python
df_agg['user_id'].nunique()
```

```python
df_agg.shape
```

**3. Feature Engineering**

We see that our full dataframe has `68807` entries/rows. However, there are so many duplicate `user_id`s, since the data is for multiple weeks. We need to do some feature engineering to tidy up the data. 

**3.1. Extracting Adopted Users**

Defining  an  "adopted  user"   as  a  user  who   has  logged  into  the  product  on  three  separate
days  in  at  least  one  seven-day  period ,  identify  which  factors  predict  future  user
adoption.

```python
adopted_user = df_agg[df_agg['visited'] >= 3]
adopted_user.head()
```

Since the definition of "adopted user" is someone who has logged in 3 times in a period of 7 days (a week) at least once, we don't need to repeat the information for a given user if we have already marked them as an adopted user. Therefore, I define a new dataframe with only unique `user_id`s for adopted users.

```python
adopted_user_unique = adopted_user.drop_duplicates(subset='user_id')
adopted_user_unique
```

```python
print('percentage of users adopted to the program :', round(adopted_user_unique.shape[0]/df_agg['user_id'].nunique() * 100 , 2), '%')
```

**3.2. Re-merging**

We have now created a dataframe that has information about unique users that are marked "adopted users". We need to do a similar thing for non-adopted users and then merge the two dataframes. Again, since marking a user as adopted/non-adopted is a binary classification, we only need to keep unique user ids for each group once.

I pull the unique `user_id`s from the `adopted_user_unique` dataframe. I use this to filter the `user_id`s in the original aggregated dataframe (`df_agg`) to be left with "non-adopted" users. Then, I remove the duplicates in the new dataframe. And finally concatenate the two dataframes. 

```python
adopted_user_ids = adopted_user['user_id'].unique()
```

```python
non_adopted_user = df_agg[~df_agg['user_id'].isin(adopted_user_ids)]
```

```python
non_adopted_user_unique = non_adopted_user.drop_duplicates(subset='user_id')
```

```python
non_adopted_user['user_id'].nunique()
```

```python
non_adopted_user_unique.shape
```

```python
df_final = pd.concat([non_adopted_user_unique, adopted_user_unique], axis = 0)
df_final.head()
```

```python
df_final['user_id'].nunique()
```

As a sanity check, we need the number of unique users in the resulted dataframe to match the number of those in the original dataframe. 

```python
df_final['user_id'].nunique() == df['user_id'].nunique()
```

**3.2. Dropping unnecessary columns**

There are some columns that are inconsequential to the analysis, e.g. name of the user, or organization id, etc. I will drop them here. But before that, I first define a new column called `invited` to replace `invited_by_user_id`, as the former can have effects on how much the user logs in. If they are invited by someone else, they might be more inclined to check. So, I define this column to have binary information on whether the user was invited (1) or not (0).

```python
df_final.columns
```

```python
df_final['invited'] = df_final['invited_by_user_id'].apply(lambda x: 0 if pd.isnull(x) else 1)
```

Next, I create a similar column to encode whether the user is an adopted user or not. This is going to be our **dependent** variable for modeling. 

```python
df_final['adopted_user'] = df_final['visited'].apply(lambda x : 1 if x >=3 else 0)
```

```python
df_final.head()
```

```python
df_final.drop(columns = ['time_stamp', 'last_session_creation_time', 'org_id', 'invited_by_user_id', 'visited'], axis = 1, inplace = True)

df_final.head()
```

```python
df_final.set_index('user_id', inplace = True)
```

**3.3. Categporical $\rightarrow$ Numeric Conversion**

```python
df_final = pd.get_dummies(df_final, drop_first = True)
```

```python
df_final
```

**4. Visualization**

```python
sns.countplot(x = 'adopted_user',
              data = df_final)
plt.grid(False)
plt.xticks([0,1])
plt.xlabel('User Types')
plt.ylabel('Count')
plt.title('Count of each user type')
plt.show();
```

```python
sns.pairplot(df_final, corner = True)
plt.savefig('pairplot.png')
plt.show();
```

The barplot is not so clear in terms of correlation between the independent variables and the dependent one. Let's see if a heatmap is going to give us a better visualization.

```python
import numpy as np
sns.heatmap(df_final.corr(),
            annot = True,
            fmt=".2f",
            cmap='coolwarm',
            annot_kws={"size": 6},
            linewidths=0.8,
            mask = np.triu(df_final.corr()));
```

It is still not a very good indicator and there might actually be no linear correlation between the variables. However, the `invited` variable does not seem to play a role in the user retention. With that in mind, I will use the other two variables in the modeling below.

**5. Modeling**



**5.1. Model 1- All features**

```python
df_final.columns
```

```python
X = df_final.drop('adopted_user', axis=1)
y = df_final[['adopted_user']]
```

We know that we have misbalanced data. So, I use the `stratify` argument in the train/test split command to make sure the ratio is similar in both training and test sets. 

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25,  stratify = y)
```

```python
#sanity check
X_train.head()
```

```python
m1 = LogisticRegression()
m1.fit(X_train, y_train)
y_pred = m1.predict(X_test)
```

```python
print(classification_report(y_test, y_pred))
```

**Positive/Adopted Class**

The recall value of 0.21 means that of all actual retained customers, only 21% were correctly predicted.

The precision value of 0.95 means of all customers predicted as retained, 95% were actually adopted users.

**Negative/Non-adopted Class**

The recall value of 1.00 means that of all actual retained customers, only 100% were correctly predicted.

The precision value of 0.87 means of all customers predicted as retained, 87% were actually non-adopted users.

Since our goal here is to catch the non-adopted customers, this model is already doing a great job at that. But, let's see if we can improve upon it.

```python
coefficients = m1.coef_[0] 
features = X_train.columns

coef_df = pd.DataFrame({
    'Feature': features,
    'Coefficient': coefficients,
    'AbsValue': np.abs(coefficients)
}).sort_values(by='AbsValue', ascending=False)

# Display top features
print(coef_df[['Feature', 'Coefficient']].head(10))
```


What this means:

<div style="text-align: left;">
    
| Feature                              | Coefficient | Interpretation                                                     |
| :------------------------------------| :---------- | :------------------------------------------------------------------ |
| `invited`                            | **–0.85**   | Strong **negative** impact on adoption likelihood                  |
| `opted_in_to_mailing_list`           | +0.64       | Users who opted in are **more likely** to adopt                    |
| `creation_source_SIGNUP`             | –0.39       | Self-signups are **less likely** to adopt                          |
| `creation_source_ORG_INVITE`         | –0.35       | Surprisingly, org-invited users are **less likely** (in your data) |
| `creation_source_SIGNUP_GOOGLE_AUTH` | –0.34       | Google-auth users slightly less likely to adopt                    |
| `enabled_for_marketing_drip`         | +0.30       | Being in the drip campaign **slightly increases** adoption odds    |
| `creation_source_PERSONAL_PROJECTS`  | –0.02       | No meaningful effect                                               |

</div>
In our next model, I keep all but `creation_source_PERSONAL_PROJECTS`, which seems to have no bearing on the results 

**5.2. Model 2**


```python
X2 = df_final.drop(['adopted_user','creation_source_PERSONAL_PROJECTS'], axis=1)

y2 = df_final[['adopted_user']]

X_train_2, X_test_2, y_train_2, y_test_2 = train_test_split(X2, y2, test_size=0.25,  stratify = y2)

m2 = LogisticRegression()
m2.fit(X_train_2, y_train_2)

y_pred_2 = m2.predict(X_test_2)

print(classification_report(y_test_2, y_pred_2))
```

**6. Comparison and Model Selection**

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4)) 

#  confusion matrix for m3
cm1 = confusion_matrix(y_test, y_pred, labels=m1.classes_)
disp1 = ConfusionMatrixDisplay(confusion_matrix=cm1,
                               display_labels=['non-adopted', 'adopted'])
axes[0].set_xticks([0, 1])
axes[0].set_yticks([0, 1])
axes[0].set_xticklabels(['non-adopted', 'adopted'], fontsize=8)
axes[0].set_yticklabels(['non-adopted', 'adopted'], fontsize=8)
disp1.plot(ax=axes[0], colorbar=False)
axes[0].set_title("Model 1")

#  confusion matrix for m1
cm2 = confusion_matrix(y_test_2, y_pred_2, labels=m2.classes_)
disp2 = ConfusionMatrixDisplay(confusion_matrix=cm2,
                               display_labels=['non-adopted', 'adopted'])
axes[1].set_xticks([0, 1])
axes[1].set_yticks([0, 1])
axes[1].set_xticklabels(['non-adopted', 'adopted'], fontsize=8)
axes[1].set_yticklabels(['non-adopted', 'adopted'], fontsize=8)
disp2.plot(ax=axes[1], colorbar=False)
axes[1].set_title("Model 2")

plt.tight_layout()
plt.show()
```
